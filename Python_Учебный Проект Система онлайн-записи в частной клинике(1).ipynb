# Customer & Sales Analytics — Python Case
Сводный анализ показателей из портфолио. Построчные исходные CSV/XLSX не приложены, поэтому ниже проверяются опубликованные KPI; шаблон аудита можно запустить после добавления исходных файлов.

```python
import pandas as pd
import matplotlib.pyplot as plt

summary = pd.DataFrame({
    "metric": ["Revenue", "Cost", "Reported Profit"],
    "value": [21954196.69, 10907991.26, 10831504.03]
})
display(summary)
```

## Проверка арифметики

```python
revenue, cost = 21954196.69, 10907991.26
reported_profit, approved_transactions = 10831504.03, 19821
calculated_profit = revenue - cost
print("Calculated profit:", round(calculated_profit, 2))
print("Reported profit:", reported_profit)
print("Calculated margin (%):", round(calculated_profit / revenue * 100, 2))
print("Calculated AOV:", round(revenue / approved_transactions, 2))
```

## Визуализация финансовых показателей

```python
ax = summary.plot(kind="bar", x="metric", y="value", legend=False, figsize=(8,4))
ax.set_title("Customer & Sales — financial summary")
ax.set_ylabel("Source currency units (currency not specified)")
ax.set_xlabel("")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## RFM — Champions

```python
champions, rfm_customers = 578, 3493
champions_profit = 2801992.29
print("Champions share (%):", round(champions / rfm_customers * 100, 2))
print("Champions profit:", f"{champions_profit:,.2f}")
```
